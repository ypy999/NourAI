```python
# 3. تجربة التوليد العصبي من النموذج بعد التدريب
from inference.engine import InferenceEngine, GenerationConfig
eng = InferenceEngine.from_checkpoint('checkpoints/v1_2/nourai_v1_2_model.pt')
for q in ['ما هي عاصمة قطر وما أهميتها؟', 'What is machine learning?', 'Quelle est la capitale de la France ?']:
    out = eng.generate(f'<|user|>\n{q}\n<|assistant|>\n', GenerationConfig(max_new_tokens=48, strategy='greedy'))
    print(q, '->', out['generated_text'])
```
