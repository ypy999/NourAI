```python3
# 3. تجربة التوليد العصبي المباشر من النموذج بعد التدريب على GPU
from inference.engine import InferenceEngine, GenerationConfig
eng = InferenceEngine.from_checkpoint('checkpoints/v1_2/nourai_v1_2_model.pt')
cfg = GenerationConfig(max_new_tokens=64, strategy='greedy', stop_sequences=('<|endoftext|>', '<|user|>'))
for q in ['ما هي عاصمة قطر وما أهميتها؟', 'مرحبا، كيف حالك؟', 'What is machine learning?', 'Quelle est la capitale de la France ?']:
    out = eng.generate(f'<|user|>\n{q}\n<|assistant|>\n', cfg)
    print(q, '->', out['generated_text'])
```
